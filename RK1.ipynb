```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
```

Громов Владислав ИУ5-63Б РК1

```python
df = pd.read_csv("Admission_Predict.csv")
print(df.isna().sum())
```

```python
corr_matrix = df.corr()


plt.figure(figsize=(10, 8))
sns.heatmap(corr_matrix, 
            annot=True,          
            cmap='coolwarm',     
            fmt=".2f",          
            vmin=-1, vmax=1)
plt.title("Матрица корреляций (Correlation Matrix)", fontsize=14)
plt.tight_layout()
plt.show()


column_to_plot = 'GRE Score'

plt.figure(figsize=(8, 6))
sns.boxplot(y=df[column_to_plot], color='skyblue')

plt.title(f"Ящик с усами (Boxplot) для признака: {column_to_plot}", fontsize=14)
plt.ylabel(column_to_plot, fontsize=12)
plt.tight_layout()
plt.show()
```

Корреляционный анализ показывает что все признаки сильно кореллируют с таргетом, видны очень сильные корреляции между признаками, что может привести к мультиколлиниарнсти

Самый сильный предиктор CPGA, самый слабый Serial № что и логично 

Потребуются модели устойчивые к мультиколлиниарности (напрмер RF или CB) либо линейные с сильно регуляризацией
